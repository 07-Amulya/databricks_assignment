```python
employee_df = spark.table("sample.default.employee_q_1")
department_df = spark.table("sample.default.department_q_1")
country_df = spark.table("sample.default.country_q_1")
```

```python
%sql
CREATE VOLUME IF NOT EXISTS workspace.default.source_to_bronze;
```

```python
(
    employee_df.write
    .format("csv")
    .option("header", True)
    .mode("overwrite")
    .save("/Volumes/sample/default/assignment_volume/source_to_bronze/employee_df")
)
```

```python
(
    department_df.write
    .format("csv")
    .option("header", True)
    .mode("overwrite")
    .save("/Volumes/sample/default/assignment_volume/source_to_bronze/department_df")
)
```

```python
(
    country_df.write
    .format("csv")
    .option("header", True)
    .mode("overwrite")
    .save("/Volumes/sample/default/assignment_volume/source_to_bronze/country_df")
)
```
