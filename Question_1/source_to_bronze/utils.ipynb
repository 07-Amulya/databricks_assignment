```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
def read_csv(path, schema):
    
    df = (
        spark.read
        .format("csv")
        .option("header", True)
        .schema(schema)
        .load(path)
    )
    
    return df
```

```python
import re

def camel_to_snake(column_name):
    
    result = ""
    
    for char in column_name:
        
        if char.isupper():
            result += "_" + char.lower()
        
        else:
            result += char
    
    return result.lstrip("_")
```

```python
camel_to_snake_udf = udf(camel_to_snake, StringType())
```

```python
def convert_columns_to_snake_case(df):
    
    new_columns = []
    
    for column_name in df.columns:
        new_columns.append(camel_to_snake(column_name))
    
    for old_column, new_column in zip(df.columns, new_columns):
        df = df.withColumnRenamed(old_column, new_column)
    
    return df
```

```python
def add_load_date(df):
    
    return df.withColumn(
        "load_date",
        current_date()
    )
```

```python
def write_delta(df, path, mode="overwrite"):
    
    (
        df.write
        .format("delta")
        .mode(mode)
        .save(path)
    )
```
